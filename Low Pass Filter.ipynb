```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# 1. Simulate data (Include High Frequency Noise)
t = np.linspace(0, 1, 1000, endpoint=False)
# Initialize Low Frequency Data (e.g. 5Hz)
clean_signal = np.sin(2 * np.pi * 5 * t)
# Add High freuqency noise
noise = 0.3 * np.random.normal(size=t.shape)
raw_data = clean_signal + noise

# 2. Set up Butterworth Low-Pass Filter
# Cut-off frequency
# H(z) is from H(s) of Laplace Transform
cutoff_freq = 0.05 # Unify frequency (From 0 to 1)
b, a = signal.butter(4, cutoff_freq, btype='low', analog=False)

# b, a represents numerator and denominator of z-transform respectively
# H(z) = (b0 + b1*z^-1 + ...) / (1 + a1*z^-1 + ...)

# 3. Apply Filter (Undergo finite difference)
filtered_data = signal.lfilter(b, a, raw_data)

# 4. Graph Comparison
plt.figure(figsize=(12, 6))
plt.plot(t, raw_data, alpha=0.5, label='Raw Data (with Noise)')
plt.plot(t, filtered_data, color='red', linewidth=2, label='Filtered Data (z-transform Logic)')
plt.title("Applying Digital Filter (Discrete System Modeling)")
plt.legend()
plt.grid(True)
plt.show()
```
