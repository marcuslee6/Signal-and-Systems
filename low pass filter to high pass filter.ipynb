```python
import numpy as np
from scipy import signal

# 1. Design 2 step IIR low pass filter，output as ZPK (zeroes, poles, gain) format
N = 2
Wn_low = 0.2 # low pass cut-off frequency
z_low, p_low, k_low = signal.butter(N, Wn_low, btype='low', output='zpk')

# 2. Use lp2hp of SciPy function to change from low pass filter to high pass filter
# high pass filter cut off frequency 0.3
Wn_high = 0.3
z_high, p_high, k_high = signal.lp2hp_zpk(z_low, p_low, k_low, wo=Wn_high)

b_high_iir, a_high_iir = signal.zpk2tf(z_high, p_high, k_high)

print("--- IIR transformed ---")
print("new high pass filter coefficient of b numerator: ", b_high_iir)
print("new high pass filter coefficient of a denominator: ", a_high_iir)
```
